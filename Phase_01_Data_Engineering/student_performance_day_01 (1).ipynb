```python
import pandas as pd


df = pd.read_csv("student_performance.csv")

df["total_score"] = df["math_score"] + df["science_score"] + df["english_score"] + df["programming_score"]

print("=" * 60)
print("      STUDENT PERFORMANCE ANALYSIS USING PANDAS")
print("=" * 60)

print("\nTOTAL NO OF COLUMNS :", len(df.columns))
print("TOTAL NO OF STUDENTS :", len(df))
print("NO OF UNIQUE DEPARTMENTS :", df['department'].nunique())

print("\n" + "=" * 60)
print("COUNT OF STUDENTS IN EACH DEPARTMENT")
print("=" * 60)

dept_count = df['department'].value_counts()

for dept, count in dept_count.items():
    print(dept, ":", count)

print("\n" + "=" * 60)
print("MATH SCORE ANALYSIS")
print("=" * 60)

print("HIGHEST MATH SCORE :", df['math_score'].max())
print("LOWEST MATH SCORE :", df['math_score'].min())
print("AVERAGE MATH SCORE :", round(df['math_score'].mean(), 2))

print("\nAVERAGE PROGRAMMING SCORE :", round(df['programming_score'].mean(), 2))

print("AVERAGE ATTENDANCE PERCENTAGE :", round(df['attendance_percentage'].mean(), 2))

print("\n" + "=" * 60)
print("AVERAGE TOTAL SCORE GROUPED BY GENDER")
print("=" * 60)

gender_avg = df.groupby('gender')['total_score'].mean()

for gender, avg in gender_avg.items():
    print(gender, ":", round(avg, 2))

print("\n" + "=" * 60)
print("TOP 5 PERFORMERS")
print("=" * 60)

top5 = df.sort_values(by='total_score', ascending=False).head(5)

print(top5[['name', 'department', 'total_score']])

print("\n" + "=" * 60)
print("STUDENTS WITH ATTENDANCE BELOW 75%")
print("=" * 60)

low_attendance = df[df['attendance_percentage'] < 75]

if len(low_attendance) > 0:
    print(low_attendance[['name', 'department', 'attendance_percentage']])
else:
    print("All students have attendance above 75%")
```
